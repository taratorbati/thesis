# SAC Irrigation Training — v2.7.0 (Colab Pro)

**Architecture:** CTDE + VDN Factorized Critic  
**Hyperparameters:** `ent_coef=0.05` (fixed), `max_grad_norm=1.0`, LR decay

## Key changes from v2.6
| # | Change | Effect |
|---|--------|--------|
| 1 | gamma obs slot restored to `elev_norm` | Bug fix: was writing field-uniform GDD scalar instead of per-agent elevation |
| 2 | 3 new static topo features per agent | Actor can now differentiate agents by cascade role, not just soil moisture |
| 3 | OBS_DIM 707 → 1097 | Per-agent block grows from 5 to 8 features |
| 4 | Episodes always run 93 days | Agent now feels late-season drought after overspending; no early termination |
| 5 | Reward simplified to r1+r2+r3+r6 | Burn-rate `rb` and dead delta-u `r5` removed |

```python
# ── CELL 6: Post-training diagnostics (v2.7)  ───────────────────────────────
#
# Runs a single wet/100% episode with the best_model and computes:
#
#   spatial_std          > 0.01  (v2.6 was ~0.002 — spatially blind)
#   corr(u, rain_today)  < 0     (v2.6 was +0.38 — irrigated MORE when raining)
#   corr(u, x1)_pooled   < 0     (v2.6 was ~0; POOLED across all days × agents)
#   ep_len               == 93
#
# IMPORTANT — correlation methodology (v2.8.1 fix):
#   corr(u, x1) is computed on the POOLED (day × agent) dataset:
#     u_pool   shape (93 × 130,) = irrigation applied by each agent each day
#     x1_pool  shape (93 × 130,) = corresponding soil moisture
#   This gives N=12 090 paired observations — statistically meaningful.
#   The previous approach (field-mean u vs field-mean x1 over time, N=93)
#   had too much variance to distinguish signal from noise.
#
#   corr(u, rain) is still computed on the field-mean daily time-series
#   (N=93) because rain is a field-uniform scalar — pooling by agent adds
#   no information.
#
# Obs normalisation: v2.8.1 fix is in runner.py.  The diagnostic here reads
# obs from the eval env (gym_env._build_obs) so it is unaffected by the fix.
# The fix only matters for the evaluation scenario results (exp_rl.py).

import numpy as np
from stable_baselines3 import SAC
from stable_baselines3.common.vec_env import DummyVecEnv
from src.rl.gym_env import IrrigationEnv, _FC_MM, _WP_MM

# ── load model ────────────────────────────────────────────────────────────────
best_model_path = f'/content/thesis/results/rl/sac_seed{SEED}_v27_*/best_model/best_model'
import glob
paths = sorted(glob.glob(best_model_path))
assert paths, f"No best_model found matching {best_model_path}"
best_model_path = paths[-1]
print(f"Loading: {best_model_path}")

# ── run wet/100% episode ──────────────────────────────────────────────────────
def make_wet_100_env():
    env = IrrigationEnv(randomize=False, curriculum_warmup_steps=0)
    # Override year to wet (2024) and budget to 100%
    env._override_year   = 2024
    env._override_budget = 484.0
    return env

# Patch: subclass or monkeypatch to force wet year
# Simpler: use randomize=False which gives dry/100, then override post-reset
eval_env = DummyVecEnv([lambda: IrrigationEnv(randomize=False, curriculum_warmup_steps=0)])
loaded   = SAC.load(best_model_path, env=eval_env)

obs = eval_env.reset()
# After reset, manually patch the env's internal state to wet year
# (requires access to the underlying env)
try:
    from climate_data import load_cleaned_data, extract_scenario
    from soil_data import get_crop
    from src.precompute import get_precomputed
    inner = eval_env.envs[0]
    df   = load_cleaned_data()
    crop = get_crop('rice')
    inner._year       = 2024
    inner._budget_mm  = 484.0
    inner._climate    = extract_scenario(df, 2024, crop)
    inner._precomp    = get_precomputed('wet', 'rice')
    obs = eval_env.reset()   # re-reset to get consistent initial obs
    inner._year       = 2024
    inner._budget_mm  = 484.0
    inner._climate    = extract_scenario(df, 2024, crop)
    inner._precomp    = get_precomputed('wet', 'rice')
except Exception as e:
    print(f"Warning: could not force wet year ({e}); running with default (dry).")

# ── collect episode ───────────────────────────────────────────────────────────
# Store per-step: (130,) action array and (130,) x1 array for pooled correlation
all_actions_per_agent = []   # list of (130,) arrays — one per day
all_x1_per_agent      = []   # list of (130,) arrays — one per day
all_rain_daily        = []   # list of float — one per day (field-uniform)
ep_len = 0
done   = False

while not done:
    action, _ = loaded.predict(obs, deterministic=True)
    all_actions_per_agent.append(action[0].copy())        # (130,)
    raw_obs = obs[0]                                       # (1097,) v2.7
    # Per-agent block: 8 features × 130 agents, agent-major
    agent_grid = raw_obs[:1040].reshape(130, 8)
    # Feature 0 is x1_norm = (x1 - WP)/(FC - WP), so x1 = x1_norm*(FC-WP)+WP
    x1_norm_agents = agent_grid[:, 0]
    x1_agents = x1_norm_agents * (_FC_MM - _WP_MM) + _WP_MM
    all_x1_per_agent.append(x1_agents)
    # Rain is scalar block index 4 (after 1040 per-agent features)
    all_rain_daily.append(float(raw_obs[1044]))
    obs, reward, done, info = eval_env.step(action)
    ep_len += 1
    done = done[0]

# ── compute metrics ───────────────────────────────────────────────────────────
actions_arr = np.array(all_actions_per_agent)   # (T, 130) in [0,1]
x1_arr      = np.array(all_x1_per_agent)        # (T, 130)
rain_arr    = np.array(all_rain_daily)           # (T,)
T, N        = actions_arr.shape
u_mm_arr    = actions_arr * 12.0                # convert to mm/day

# Field-mean time series
u_mean_daily = u_mm_arr.mean(axis=1)            # (T,)

# Spatial std (how much does daily allocation vary across agents?)
spatial_std = actions_arr.mean(axis=0).std()

# corr(u, rain): field-mean daily, N=T=93
corr_rain = float(np.corrcoef(u_mean_daily, rain_arr)[0, 1])

# corr(u, x1): POOLED across all (day, agent) pairs, N = T*130 = 12090
# This is statistically robust; the N=93 time-series version is not.
u_pool  = u_mm_arr.flatten()   # (T*130,)
x1_pool = x1_arr.flatten()     # (T*130,)
corr_x1_pooled = float(np.corrcoef(u_pool, x1_pool)[0, 1])

# waterlog days: x1 > FC = 140mm
waterlog_mask = x1_arr > _FC_MM
waterlog_days = waterlog_mask.mean(axis=0).mean() * T   # per-agent mean days

print('=== v2.7 Post-training Diagnostics (wet/100% episode) ===')
print(f'Episode length:              {ep_len}  (expected 93)')
print(f'Mean irrigation mm/day:      {u_mean_daily.mean():.2f}')
print(f'Spatial std (daily alloc):   {spatial_std:.4f}  (target > 0.01; v2.6 was ~0.002)')
print(f'corr(u_mean, rain_today):    {corr_rain:+.3f}  (target < 0; v2.6 was +0.38)')
print(f'corr(u, x1) POOLED NxT:     {corr_x1_pooled:+.3f}  (target < 0; v2.6 was ~+0.03)')
print(f'  (pooled N = {T*N:,} = {T} days x {N} agents)')
print(f'Waterlog days/agent:         {waterlog_days:.1f}  (MPC is ~20; v2.6 was ~90)')
print()
checks = [
    (ep_len == 93,          f'ep_len {ep_len} == 93'),
    (spatial_std > 0.01,    f'spatial_std {spatial_std:.4f} > 0.01'),
    (corr_rain < 0,         f'corr(u,rain) {corr_rain:+.3f} < 0'),
    (corr_x1_pooled < 0,    f'corr(u,x1) pooled {corr_x1_pooled:+.3f} < 0'),
    (waterlog_days < 60,    f'waterlog_days {waterlog_days:.1f} < 60'),
]
for ok, label in checks:
    print(f'  {"v" if ok else "!"} {label}')
```
